```python
import pandas as pd
```

```python
df_raw = pd.read_csv('Absenteeism_data.csv')
df_raw.head()
```

```python
pd.options.display.max_columns = None
pd.options.display.max_rows = 20
```

```python
df = df_raw.copy()
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df = df.drop(['ID'], axis=1)
df.head()
```

```python
sorted(df["Reason for Absence"].unique())
```

```python
df.head()
```

```python
reason_cols = pd.get_dummies(df["Reason for Absence"],drop_first=True)
reason_cols
```

```python
reason_cols["Check"] = reason_cols.sum(axis=1)
reason_cols
```

```python
reason_cols[reason_cols["Check"] != 1]
```

```python
reason_cols = reason_cols.drop(["Check"], axis=1)
reason_cols.head()
```

```python
df.head()
```

```python
df = df.drop([ 'Reason for Absence'], axis=1)
```

```python
reason_type_1 = reason_cols.loc[:, 1:14].max(axis=1)
reason_type_2 = reason_cols.loc[:, 15:17].max(axis=1)
reason_type_3 = reason_cols.loc[:, 18:21].max(axis=1)
reason_type_4 = reason_cols.loc[:, 22:].max(axis=1)
```

```python
df = pd.concat([df, reason_type_1, reason_type_2, reason_type_3, reason_type_4], axis=1)
df.head()
```

```python
df.columns.values
```

```python
col_names = ['Date', 'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education',
       'Children', 'Pets', 'Absenteeism Time in Hours', "Reason_1", "Reason_2", "Reason_3", "Reason_4"]
df.columns = col_names
df.head()
```

```python
age_dummies = pd.get_dummies(df["Age"], drop_first=True)
age_dummies.head()
```

```python
df.head()
```

```python
df_reason_mod = df.copy()
```

```python
df_reason_mod["Date"] = pd.to_datetime(df_reason_mod["Date"], format="%d/%m/%Y")
df_reason_mod["Month"] = df_reason_mod["Date"].dt.month
df_reason_mod["Day of Week"] = df_reason_mod["Date"].dt.dayofweek
df_reason_mod.head()
```

```python
df_reason_mod.drop(["Date"], axis=1, inplace=True)
df_reason_mod.head()
```

```python
df_reason_date_mod = df_reason_mod.copy()
df_reason_date_mod.head()
```

```python
df_reason_date_mod["Education"].unique()
```

```python
df_reason_date_mod["Education"] = df_reason_date_mod["Education"].map({1:0, 2:1, 3:1, 4:1})
df_reason_date_mod["Education"].unique()
```

```python
df_reason_date_mod.columns.values
```

```python
col_names = [ 'Reason_1',
       'Reason_2', 'Reason_3', 'Reason_4', 'Month', 'Day of Week', 'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education',
       'Children', 'Pets', 'Absenteeism Time in Hours',]
```

```python
df_reason_date_mod = df_reason_date_mod[col_names]
df_reason_date_mod.head()
```

```python
df_preprocessed = df_reason_date_mod.copy()
df_preprocessed.head()
```

```python
df_preprocessed.to_csv('Absenteeism_preprocessed.csv', index=False)
```
